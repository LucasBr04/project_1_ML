# BRFSS 2015 — Least Squares et Ridge Regression

We compare Least Squares and Ridge using the **F1-score for the +1 class**. We explore several values ​​of λ and visualize their effects.

**Protocol without data leakage:** `prepare_data()` generates a training set (80%) and a validation set (20%), with preprocessing fitted exclusively on the training data. We further subdivide the validation set into a **calibration** set (for selecting λ and the threshold) and a **final evaluation** set (which is never used for parameter selection). The competition test set is not used here.

## 1. Imports et configuration

```python
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path
from preprocessing import load_feature_names
from implementations import (
    least_squares,
    ridge_regression
)
from helpers import load_csv_data
from preprocessing import load_feature_names, clean_and_standardize

SEED = 42
import csv

Path("plot").mkdir(exist_ok=True)


def show_table(rows):
    """Print an aligned table without changing the underlying values."""
    if not rows:
        return

    columns = list(rows[0])

    def format_value(value):
        if isinstance(value, (float, np.floating)):
            if np.isnan(value):
                return "—"
            if value != 0 and abs(value) < 1e-3:
                return f"{value:.4e}"
            return f"{value:.4f}"
        return str(value)

    formatted_rows = [
        [format_value(row[column]) for column in columns]
        for row in rows
    ]

    widths = [
        max(len(str(column)), max(len(row[i]) for row in formatted_rows))
        for i, column in enumerate(columns)
    ]

    def format_row(values):
        return " | ".join(
            value.ljust(width) if i == 0 else value.rjust(width)
            for i, (value, width) in enumerate(zip(values, widths))
        )

    print(format_row([str(column) for column in columns]))
    print("-+-".join("-" * width for width in widths))

    for row in formatted_rows:
        print(format_row(row))


def save_table(path, rows, index_label=None):
    """Write records with the same column order as the original CSV tables."""
    columns = list(rows[0])
    with open(path, "w", newline="", encoding="utf-8") as stream:
        writer = csv.writer(stream)
        writer.writerow(([""] if index_label is not None else []) + columns)
        for i, row in enumerate(rows):
            values = [row[k] for k in columns]
            values = ["" if isinstance(value, (float, np.floating)) and np.isnan(value)
                      else value for value in values]
            writer.writerow(([i] if index_label is not None else []) + values)


def table_column(rows, key):
    return np.array([row[key] for row in rows])


def pivoted_qr_numpy(matrix):
    """Return R and column pivots using only NumPy.

    First reduce the tall input with an orthogonal QR transformation. Column
    pivoting then uses Householder reflections on the small reduced matrix.
    Orthogonal reduction preserves column norms and linear dependence, without
    forming X.T @ X (which would square the condition number).
    """
    r = np.linalg.qr(np.asarray(matrix, dtype=float), mode="r")
    pivots = np.arange(r.shape[1])
    for k in range(min(r.shape)):
        norms = np.linalg.norm(r[k:, k:], axis=0)
        pivot = k + int(np.argmax(norms))
        r[:, [k, pivot]] = r[:, [pivot, k]]
        pivots[[k, pivot]] = pivots[[pivot, k]]
        vector = r[k:, k].copy()
        norm = np.linalg.norm(vector)
        if norm == 0:
            continue
        alpha = -np.copysign(norm, vector[0])
        vector[0] -= alpha
        vector /= np.linalg.norm(vector)
        block = r[k:, k:]
        block -= 2 * np.outer(vector, vector @ block)
        r[k, k] = alpha
        r[k + 1:, k] = 0
    return r, pivots
```

## 2. Préparation et vérification des classes
Preprocessing handles imputation, encoding, and standardization. No observations are removed.

```python
# Load data
data_path = "data"
x, x_test, y, _, test_ids = load_csv_data(data_path)

# Load feature names
names = load_feature_names(data_path)

# Split train / validation (80% / 20%)
np.random.seed(42)
indices = np.random.permutation(len(y))

split = int(0.8 * len(y))

train_idx = indices[:split]
val_idx = indices[split:]

X_train_raw = x[train_idx]
X_val_raw = x[val_idx]

y_train = y[train_idx]
y_val = y[val_idx]

X_train, X_val = clean_and_standardize(
    X_train_raw,
    X_val_raw,
    names
)
```

## 3. Diviser la validation en calibration et évaluation
As explained previously, `X_cal` and `Y_cal` will serve as data for calibration and for finding the optimal `lambda` and `threshold` values. Subsequently, `X_eval` and `Y_eval` will be used for the final evaluation (F1 score, recall, and precision).

```python
def stratified_indices(y, fraction=0.5, seed=123):
    rng = np.random.default_rng(seed)
    left, right = [], []
    for label in np.unique(y):
        idx = rng.permutation(np.flatnonzero(y == label))
        n_left = int(round(len(idx) * fraction))
        left.extend(idx[:n_left]); right.extend(idx[n_left:])
    return rng.permutation(left), rng.permutation(right)

cal_idx, eval_idx = stratified_indices(y_val, fraction=0.5, seed=SEED + 1)
X_cal, y_cal = X_val[cal_idx], y_val[cal_idx]
X_eval, y_eval = X_val[eval_idx], y_val[eval_idx]
print("Calibration :", X_cal.shape, "positifs :", np.mean(y_cal == 1))
print("Evaluation :", X_eval.shape, "positifs :", np.mean(y_eval == 1))
```

## 4. Métriques adaptées au déséquilibre

**Important:** LS/Ridge models predict a **continuous score**, not directly −1/+1. We classify an instance as `+1` if `score >= threshold`, and as `−1` otherwise. A threshold of 0 is natural for −1/+1 targets, but it is **not necessarily optimal for the F1 score**.

```python
def predict_labels(scores, threshold=0.0):
    return np.where(np.asarray(scores) >= threshold, 1, -1)


def classification_metrics(y_true, y_pred):
    y_true = np.asarray(y_true)
    y_pred = np.asarray(y_pred)
    tp = int(np.sum((y_true == 1) & (y_pred == 1)))
    fp = int(np.sum((y_true == -1) & (y_pred == 1)))
    fn = int(np.sum((y_true == 1) & (y_pred == -1)))
    tn = int(np.sum((y_true == -1) & (y_pred == -1)))
    precision = tp / (tp + fp) if tp + fp else 0.0
    recall = tp / (tp + fn) if tp + fn else 0.0
    f1 = 2 * tp / (2 * tp + fp + fn) if 2 * tp + fp + fn else 0.0
    accuracy = (tp + tn) / len(y_true)
    return dict(F1=f1, Precision=precision, Recall=recall,
                Accuracy=accuracy, TP=tp, FP=fp, FN=fn, TN=tn)


def best_f1_threshold(y_true, scores):
    """Exact search for the optimal F1 threshold on the calibration set"""
    y_true = np.asarray(y_true)
    scores = np.asarray(scores, dtype=float)
    order = np.argsort(-scores, kind="stable")
    s = scores[order]
    positive = (y_true[order] == 1).astype(int)
    tp = np.cumsum(positive)
    fp = np.arange(1, len(s) + 1) - tp
    total_pos = positive.sum()
    f1 = 2 * tp / np.maximum(2 * tp + fp + total_pos - tp, 1)
    valid_end = np.r_[s[:-1] > s[1:], True]
    eligible = np.flatnonzero(valid_end)
    best = eligible[np.argmax(f1[eligible])]
    return float(s[best]), float(f1[best])

baseline = classification_metrics(y_eval, np.full(len(y_eval), -1))
print("Baseline always -1 :", baseline)
```

## 5. Least Squares

```python
import numpy as np

# identify columns which are linearly dependant
R, piv = pivoted_qr_numpy(X_train)
# Same SVD rank criterion as matrix_rank(X_train), using its original shape.
singular_values = np.linalg.svd(R, compute_uv=False)
tolerance = singular_values.max() * max(X_train.shape) * np.finfo(float).eps
rank = int(np.sum(singular_values > tolerance))

independent_cols = piv[:rank]
dependent_cols = piv[rank:]

#print("Independent columns:", independent_cols)
#print("Dependent columns:", dependent_cols)

# delete dependant columns
X_train_fixed = X_train[:, independent_cols]
X_cal_fixed = X_cal[:, independent_cols]
X_eval_fixed = X_eval[:, independent_cols]

# Least Squares
w_ls, ls_train_loss = least_squares(
    y_train.astype(float),
    X_train_fixed
)

# Compute scores
ls_cal_scores = X_cal_fixed @ w_ls
ls_eval_scores = X_eval_fixed @ w_ls

# Search best threshold for callibration
ls_threshold, ls_cal_f1 = best_f1_threshold(
    y_cal,
    ls_cal_scores
)

#  Evaluate with default threshold
ls_eval_default = classification_metrics(
    y_eval,
    predict_labels(ls_eval_scores, 0.0)
)

# evaluate with optimal threshold
ls_eval_tuned = classification_metrics(
    y_eval,
    predict_labels(ls_eval_scores, ls_threshold)
)

# print results
print(
    f"Least Squares : demi-MSE train = {ls_train_loss:.5f} ; "
    f"seuil calibré = {ls_threshold:.5f}"
)

ls_rows = [
    {"Model": "LS, threshold 0", **ls_eval_default},
    {"Model": "LS, callibrate threshold", **ls_eval_tuned},
]
show_table(ls_rows)
# Preserve the four-decimal rounding used by the original df_ls.csv.
rounded_ls_rows = [{key: round(value, 4) if isinstance(value, float) else value
                    for key, value in row.items()} for row in ls_rows]
save_table("plot/df_ls.csv", rounded_ls_rows)
```

## 6. Ridge Regression


```python
n_train, n_features = X_train.shape
ridge_gram = X_train.T.dot(X_train)
ridge_rhs = X_train.T.dot(y_train.astype(float))
lambdas = np.logspace(-6, 2, 30)
results = []
ridge_weights = {}
ridge_thresholds = {}

for lam in lambdas:
    # Exactly equivalent to ridge_regression(y_train, X_train, lam).
    w = np.linalg.solve(
        ridge_gram + 2 * n_train * lam * np.identity(n_features),
        ridge_rhs,
    )
    ridge_weights[float(lam)] = w
    cal_scores = X_cal @ w
    threshold, cal_f1 = best_f1_threshold(y_cal, cal_scores)
    ridge_thresholds[float(lam)] = threshold
    at_zero = classification_metrics(y_cal, predict_labels(cal_scores, 0.0))
    tuned = classification_metrics(y_cal, predict_labels(cal_scores, threshold))

    results.append({
        "lambda": float(lam),
        "F1_cal_threshold_0": at_zero["F1"],
        "F1_cal_optimized": cal_f1,
        "precision_cal": tuned["Precision"],
        "recall_cal": tuned["Recall"],
        "threshold": threshold,
        "weight_norm": np.linalg.norm(w),
        "MSE_cal": np.mean((y_cal - X_cal @ w) ** 2),
    })

ridge_df = results
best_row = max(ridge_df, key=lambda row: row["F1_cal_optimized"])
best_lambda = float(best_row["lambda"])
best_ridge_threshold = ridge_thresholds[best_lambda]

print(f"Best λ (calibration) = {best_lambda:.6g}")
print(f"Optimal threshold (calibration) = {best_ridge_threshold:.6f}")
print(f"Calibration F1 = {best_row['F1_cal_optimized']:.4f}")

show_table(ridge_df)
save_table("plot/ridge_df.csv", ridge_df, index_label="")
```

## 7. Figures


```python
plt.figure(figsize=(9, 5))
plt.semilogx(
    table_column(ridge_df, "lambda"),
    table_column(ridge_df, "F1_cal_threshold_0"),
    "o-",
    label="F1, threshold 0",
)
plt.semilogx(
    table_column(ridge_df, "lambda"),
    table_column(ridge_df, "F1_cal_optimized"),
    "o-",
    label="F1, calibrated threshold",
)
plt.axvline(best_lambda, linestyle="--", label=f"Best λ = {best_lambda:.3g}")
plt.xlabel("λ (log scale)")
plt.ylabel("Positive-class F1")
plt.title("Ridge: F1 versus λ (calibration)")
plt.legend()
plt.grid(alpha=0.3)
plt.savefig("plot/ridge_f1_vs_lambda_calibration.png", bbox_inches="tight")
plt.show()
```

```python
plt.figure(figsize=(9, 5))
plt.semilogx(
    table_column(ridge_df, "lambda"),
    table_column(ridge_df, "precision_cal"),
    "o-",
    label="Precision",
)
plt.semilogx(
    table_column(ridge_df, "lambda"),
    table_column(ridge_df, "recall_cal"),
    "o-",
    label="Recall",
)
plt.xlabel("λ (log scale)")
plt.ylabel("Score")
plt.title("Ridge: precision–recall trade-off (calibrated threshold)")
plt.legend()
plt.grid(alpha=0.3)
plt.savefig(
    "plot/ridge_precision_recall_calibrated_threshold.png",
    bbox_inches="tight",
)
plt.show()
```

```python
# Additional diagnostic for the report: impact of the threshold on F1.
# Thresholds are assessed on calibration data only, without tuning on evaluation data.
cal_scores_best = X_cal @ ridge_weights[best_lambda]
threshold_grid = np.linspace(np.min(cal_scores_best), np.max(cal_scores_best), 150)
f1_by_threshold = [
    classification_metrics(y_cal, predict_labels(cal_scores_best, t))["F1"]
    for t in threshold_grid
]

plt.figure(figsize=(9, 5))
plt.plot(threshold_grid, f1_by_threshold, label="Calibration F1")
plt.axvline(
    best_ridge_threshold,
    linestyle="--",
    label=f"Selected threshold = {best_ridge_threshold:.3f}",
)
plt.axvline(0, linestyle=":", label="Default threshold = 0")
plt.xlabel("Decision threshold")
plt.ylabel("Positive-class F1")
plt.title("Ridge: F1 sensitivity to the classification threshold")
plt.legend()
plt.grid(alpha=0.3)
plt.savefig("ridge_f1_threshold_sensitivity.png", bbox_inches="tight")
plt.show()
```

## 8. Comparaison finale sur l’ensemble d’évaluation

**Une seule évaluation** après sélection sur calibration. Nous comparons les deux modèles au seuil 0 et à leur seuil calibré.

```python
best_w = ridge_weights[best_lambda]
ridge_eval_scores = X_eval @ best_w

rows = [
    {
        "Model": "Always -1 baseline",
        "lambda": np.nan,
        "threshold": np.nan,
        **baseline,
    },
    {
        "Model": "Least Squares / threshold 0",
        "lambda": 0.0,
        "threshold": 0.0,
        **ls_eval_default,
    },
    {
        "Model": "Least Squares / calibrated threshold",
        "lambda": 0.0,
        "threshold": ls_threshold,
        **ls_eval_tuned,
    },
    {
        "Model": "Ridge / threshold 0",
        "lambda": best_lambda,
        "threshold": 0.0,
        **classification_metrics(
            y_eval, predict_labels(ridge_eval_scores, 0.0)
        ),
    },
    {
        "Model": "Ridge / calibrated threshold",
        "lambda": best_lambda,
        "threshold": best_ridge_threshold,
        **classification_metrics(
            y_eval, predict_labels(ridge_eval_scores, best_ridge_threshold)
        ),
    },
]

comparison = rows
show_table(comparison)
save_table("plot/comparison.csv", comparison)
```

```python
fig, ax = plt.subplots(figsize=(9, 5))

model_names = [
    "Least Squares / threshold 0",
    "Least Squares / calibrated threshold",
    "Ridge / threshold 0",
    "Ridge / calibrated threshold",
]
by_name = {row["Model"]: row for row in comparison}
subset = [by_name[name] for name in model_names]

ax.bar(np.arange(len(subset)), table_column(subset, "F1"))
ax.set_xticks(
    np.arange(len(subset)),
    ["LS / threshold 0", "LS / calibrated", "Ridge / threshold 0", "Ridge / calibrated"],
)
ax.set_ylabel("Positive-class F1")
ax.set_title("F1 comparison on the independent evaluation set")
ax.grid(axis="y", alpha=0.3)

plt.tight_layout()
plt.show()
```
